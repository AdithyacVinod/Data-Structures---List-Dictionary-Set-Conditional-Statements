```python
# 1.List Creation

# Creating a list of five integer elements
age_list = [24, 25, 26, 27, 28]

# Creating a list of five string elements
name_list = ["Anu", "Meera", "Rahul", "Arun", "Priya"]

print("Age List:", age_list)
print("Name List:", name_list)
```

A list is used to store multiple values in a single variable. Lists are ordered and can contain different types of data.

```python
# 2.List Operations / Modifications

age_list = [24, 25, 26, 27, 28]
name_list = ["Anu", "Meera", "Rahul", "Arun", "Priya"]

# a. Append "Yazhini"
name_list.append("Yazhini")

# b. Insert 30 at index 2
age_list.insert(2, 30)

# c. Remove "Yazhini"
name_list.remove("Yazhini")

# d. Pop the last element
age_list.pop()

# e. Extend age_list
age_list.extend([29, 30, 26])

# f. Sort age_list in descending order
age_list.sort(reverse=True)

# g. Find maximum, minimum and sum
max_age = max(age_list)
min_age = min(age_list)
total_age = sum(age_list)

print("Name List:", name_list)
print("Age List:", age_list)
print("Maximum Age:", max_age)
print("Minimum Age:", min_age)
print("Sum of Ages:", total_age)
```



*  append() adds an element to the end of a list.
*   insert() adds an element at a specific index.
*   remove() removes a specified element.
*   pop() removes an element, here the last element.
*  extend() adds multiple elements.
*   sort(reverse=True) sorts the list in descending order.
*   max(), min() and sum() find the maximum, minimum and total values.






```python
# 3.Accessing List Elements

name_list = ["Anu", "Meera", "Rahul", "Arun", "Priya"]

# a. First element
print("First Element:", name_list[0])

# b. Last element
print("Last Element:", name_list[-1])

# c. Elements from index 2 to index 4
print("Index 2 to 4:", name_list[2:5])

# d. List in reverse order
print("Reverse Order:", name_list[::-1])
```

List indexing starts from 0. Negative indexing can be used to access elements from the end. Slicing [2:5] gets elements from index 2 up to index 4.

```python
# 4.Dictionary Creation, Modification and Access

# a. Creating a dictionary
student_marks = {
    "Anu": 75,
    "Meera": 88,
    "Rahul": 65,
    "Arun": 92,
    "Priya": 78
}

# b. Accessing the mark of a specific student
print("Rahul's Mark:", student_marks["Rahul"])

# c. Adding a new student
student_marks["Janani"] = 80

# d. Updating an existing student's mark
student_marks["Anu"] = 82

# e. Printing keys, values and items
print("Keys:", student_marks.keys())
print("Values:", student_marks.values())
print("Key-Value Pairs:", student_marks.items())
```

A dictionary stores data in key-value pairs. The key is used to access the corresponding value. keys() returns all keys, values() returns all values, and items() returns both keys and values.

```python
# 5.Sets(Operations)

# a)Creating a set
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}

print("My Set:", my_set)
```

A set stores unique elements only. Therefore, repeated values such as 'a' and 'i' are automatically removed. The output will contain each vowel only once. The order of elements in a set is not fixed.

```python
# b.Attempting to Change a Set Element
my_set = {'a', 'e', 'i', 'o', 'u'}

my_set[4] = 's'
```

Sets are unordered and do not support indexing. Therefore, we cannot directly change an element using an index such as my_set[4].

```python
# c.Create two sets
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

# d.Union and Intersection of Sets
# Union
union_set = set1.union(set2)

# Intersection
intersection_set = set1.intersection(set2)

print("Union:", union_set)
print("Intersection:", intersection_set)
```



*   Union combines all unique elements from both sets.
*   Intersection returns only the elements common to both sets.



```python
# 6.Operators & Conditional Statements

# Performance Category Program:
score = float(input("Enter your score (0 to 10): "))

if score < 0 or score > 10:
    print("Invalid score. Please enter a score between 0 and 10.")

elif score > 7:
    print("Above Average: Excellent performance! Keep up the good work.")

elif score >= 4:
    print("Average: Good effort! Keep practicing, there's room for improvement.")

else:
    print("Below Average: Need to improve your performance. Consistent practice will lead to better results.")


```

The if, elif, and else statements are used to make decisions based on conditions.


*   Score greater than 7 → Above Average
*   Score from 4 to 7 → Average
*   Score less than 4 → Below Average
*   Scores outside 0–10 → Invalid score
